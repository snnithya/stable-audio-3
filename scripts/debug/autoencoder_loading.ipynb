```python
from stable_audio_3 import AutoencoderModel
import numpy as np
import IPython.display as ipd
import torchaudio
import torch
```

# Recreate roundtrip encoding/decoding

```python
audio_sample = "/data/hai-res/shared/snnithya/sao-3/data/slakh-streamgen-preencoded-same-s/train/_sanity_check/0000000000_v0_source.wav"
audio, sr = torchaudio.load(audio_sample)
```

```python
ipd.Audio(audio.cpu(), rate=44100)
```

```python
sr
```

```python
# SAME-S

ae = AutoencoderModel.from_pretrained('same-s')
encoded_s = ae.encode(audio, sr)
decoded_s = ae.decode(encoded_s)
ipd.Audio(data=decoded_s[0].cpu().numpy(), rate=sr)
```

```python
# SAME-S

ae = AutoencoderModel.from_pretrained('same-s')
encoded_s = ae.encode(audio, sr, chunked=True)
decoded_s = ae.decode(encoded_s, chunked=True)
ipd.Audio(data=decoded_s[0].cpu().numpy(), rate=sr)
```

# Load encoded audio

```python
encoded_latent_path = '/data/hai-res/shared/snnithya/sao-3/data/babyslakh-streamgen-preencoded-same-s/0000000000.npy'
encoded_latent = np.load(encoded_latent_path)
```

```python
ae = AutoencoderModel.from_pretrained('same-s')
```

```python
encoded_latent_fp32 = torch.from_numpy(encoded_latent).float().cuda().unsqueeze(0)
encoded_latnet_bf16 = torch.from_numpy(encoded_latent).to(torch.bfloat16).cuda().unsqueeze(0)
```

```python

with torch.amp.autocast("cuda"):
    target_decoded_fp32 = ae.decode(encoded_latent_fp32)
    target_decoded_bf16 = ae.decode(encoded_latnet_bf16)
```

```python
ipd.Audio(target_decoded_fp32[0].cpu().numpy(), rate = 44100)
```

```python
ipd.Audio(target_decoded_bf16[0].cpu().numpy(), rate = 44100)
```
